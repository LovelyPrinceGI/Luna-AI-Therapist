```python
import json
import pandas as pd
from pathlib import Path

BASE = Path(r"C:\Luna-AI-Therapist\dissonance\craft_dialogue")

INPUT_FILES = {
    "baseline":   BASE / "baseline"   / "dialogue_1_full_baseline.jsonl",
    "emotion":    BASE / "emotion"    / "dialogue_1_full_emotion_online.jsonl",
    "dissonance": BASE / "dissonance" / "dialogue_1_full_dissonance_online.jsonl",
}

OUTPUT_CSV = {
    "baseline":   BASE / "baseline"   / "dialogue_1_baseline_turns.csv",
    "emotion":    BASE / "emotion"    / "dialogue_1_emotion_turns.csv",
    "dissonance": BASE / "dissonance" / "dialogue_1_dissonance_turns.csv",
}

def extract_turn_client_therapist(in_path: Path, out_path: Path):
    rows = []
    with in_path.open("r", encoding="utf-8") as f:
        for line in f:
            if not line.strip():
                continue
            rec = json.loads(line)
            rows.append({
                "turn": rec.get("turn"),
                "client": rec.get("client", ""),
                "therapist": rec.get("therapist", ""),
            })
    df = pd.DataFrame(rows)
    df.to_csv(out_path, index=False, encoding="utf-8-sig")
    print(f"[SAVE] {out_path} ({len(df)} turns)")

for method, in_path in INPUT_FILES.items():
    extract_turn_client_therapist(in_path, OUTPUT_CSV[method])
```
