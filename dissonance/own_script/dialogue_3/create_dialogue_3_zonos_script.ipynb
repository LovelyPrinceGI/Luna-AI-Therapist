# เจน zonos_script (dialogue_3)

```python
import os
import json

BASE_DIR = "/home/patsakornt/work/test"
DIALOGUE_ID = 3
DIALOGUE_DIR = os.path.join(BASE_DIR, "dissonance", "own_script", f"dialogue_{DIALOGUE_ID}")

# 1) โหลด JSON ที่มี directed_utterances
with open(os.path.join(DIALOGUE_DIR, f"dialogue_{DIALOGUE_ID}_directed_zonos.json"),
          "r", encoding="utf-8") as f:
    directed = json.load(f)

directed_utterances = directed.get("directed_utterances", [])
print("directed_utterances:", len(directed_utterances))

# 2) ถ้า utterance_text ใน JSON ตรงกับ text ใน dialogue_3.txt อยู่แล้ว
#    เราใช้ตรงจาก JSON ได้เลย (ไม่ต้องอ่าน txt ซ้ำ)
lines_out = []

for u in directed_utterances:
    text = u.get("utterance_text", "").strip()
    rule = u.get("new_utterance_rule_definition", {})
    vec = rule.get("primary_zonos_vector_value", {}) or {}
    speaking_rate = rule.get("speaking_rate")
    pitch_std = rule.get("pitch_std")

    # สร้าง attribute list: sadness="0.5" fear="0.3" ...
    attrs = []
    # ใช้ชื่อ emotion เป็น lower case เหมือนตัวอย่างเก่า
    for emotion, value in vec.items():
        attrs.append(f'{emotion.lower()}="{value}"')
    if speaking_rate is not None:
        attrs.append(f'speaking_rate="{speaking_rate}"')
    if pitch_std is not None:
        attrs.append(f'pitch_std="{pitch_std}"')

    attr_str = " ".join(attrs) if attrs else ""

    # line แบบเก่า:
    # <emotion ...>text</emotion>
    line = f'<emotion {attr_str}>{text}</emotion>'
    lines_out.append(line)

# 3) เซฟเป็น Zonos script สำหรับ dialogue_3
ZONOS_SCRIPT_DIR = os.path.join(BASE_DIR, "zonos_scripts")
os.makedirs(ZONOS_SCRIPT_DIR, exist_ok=True)
out_path = os.path.join(ZONOS_SCRIPT_DIR, f"dialogue_{DIALOGUE_ID}_zonos.txt")

with open(out_path, "w", encoding="utf-8") as f:
    f.write("\n".join(lines_out))

print("Saved Zonos script to:", out_path)
```
