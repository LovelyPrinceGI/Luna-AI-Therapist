```python
# -*- coding: utf-8 -*-
"""
สคริปต์สำหรับดาวน์โหลด CACTUS dataset จาก Hugging Face
โดยใช้ library `datasets`
"""

# ขั้นตอนที่ 1: Import library ที่จำเป็น
# 'load_dataset' คือฟังก์ชันหลักที่เราจะใช้
from datasets import load_dataset
import pandas as pd

print("กำลังเริ่มต้นดาวน์โหลด CACTUS dataset จาก Hugging Face...")
print("ขั้นตอนนี้อาจใช้เวลาสักครู่ ขึ้นอยู่กับขนาดของ dataset และความเร็วอินเทอร์เน็ตนะคะ")

try:
    # ขั้นตอนที่ 2: ใช้ฟังก์ชัน load_dataset เพื่อดาวน์โหลด
    # "LangAGI-Lab/cactus" คือชื่อ repository ของ dataset บน Hugging Face
    # trust_remote_code=True เป็นการอนุญาตให้รันโค้ดที่มากับ dataset (ซึ่งจำเป็นสำหรับบาง dataset)
    cactus_dataset = load_dataset("LangAGI-Lab/cactus", trust_remote_code=True)

    print("\n✅ ดาวน์โหลด CACTUS dataset สำเร็จแล้วค่ะ!")

    # ขั้นตอนที่ 3: ตรวจสอบข้อมูลที่ดาวน์โหลดมา
    # Dataset ที่โหลดมาจะอยู่ในรูปแบบของ Dictionary โดยมี key คือ 'train' (หรือ 'test', 'validation')
    print("\n--- โครงสร้างของ Dataset ---")
    print(cactus_dataset)

    # ลองดูตัวอย่างข้อมูล 1 record จาก 'train' set
    print("\n--- ตัวอย่างข้อมูลแรก (First Sample) ---")
    first_sample = cactus_dataset['train'][0]
    print(first_sample)

    # ลองดูเฉพาะส่วนของบทสนทนา (dialogue) ของตัวอย่างแรก
    print("\n--- ตัวอย่างบทสนทนา ---")
    print(first_sample['dialogue'])
    
    # เพื่อให้เห็นภาพรวมได้ง่ายขึ้น เราสามารถแปลงเป็น Pandas DataFrame ได้
    print("\n--- กำลังแปลงเป็น DataFrame เพื่อดูภาพรวม (5 แถวแรก) ---")
    df = cactus_dataset['train'].to_pandas()
    print(df.head())


except Exception as e:
    print(f"\n❌ เกิดข้อผิดพลาดระหว่างการดาวน์โหลด: {e}")
    print("กรุณาตรวจสอบการเชื่อมต่ออินเทอร์เน็ต หรือลองตรวจสอบว่าติดตั้ง library `datasets` ถูกต้องหรือไม่")

print("\nภารกิจสำเร็จ! ตอนนี้เรามีวัตถุดิบหลักพร้อมสำหรับ 'The Forge' แล้วค่ะ! 🔥")
```
